# Importing required Libraries
Importing numpy and pandas to read and process the data

```python
import numpy as np
import pandas as pd
```

Storing the data set in a variable 'df' and printing first few rows and columns to check if the dataset is properly loaded

```python
df = pd.read_csv('placement-dataset.csv')
df.head()
```

```python
df.shape
```

# Processing the data

Removes the city column as it is useless for our model

```python
df = df.drop("city", axis = 1)
```

```python
df.info()
```

Removes all the rows with atleast 1 null value

```python
df = df.dropna()
df.info()
```

```python
df
```

# Importing Matplotlib
We use scatter function of matplotlib to represent data as a scatter plot with x-axis = cgpa, y-axis = iq

```python
import matplotlib.pyplot as plt
plt.scatter(df['cgpa'], df['iq'], c = df['placement'])
```

# Creating Dependent and Independent variables
'x' is an independent variable whereas 'y' is a dependent variable

```python
x = df[['cgpa', 'iq']]
y = df['placement']
```

```python
x
```

```python
y
```

# Importing Important Functions
We import train_test_split function to make a split between the training set of data and testing set of the data

```python
from sklearn.model_selection import train_test_split
```

Here we use the function and assign the values of datas to variables. We've also set the test_size = 0.1 which basically means that the size of test dataset will 0.1(or 10%) of the total data

```python
x_train,x_test,y_train,y_test = train_test_split(x, y, test_size = 0.1)
```

```python
x_train.info()
x_test.info()
y_train.info()
y_test.info()
```

# Importing StandardScaler Class
We make an object scaler of the class StandardScaler which will allow us to use the functions we are provided in this class

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

We use fit_transform function the standardize the values of x_train and x_test datasets

```python
x_train = scaler.fit_transform(x_train)
x_test = scaler.fit_transform(x_test)
```

```python
x_train
```

```python
x_test
```

# Importing LogiticRegression
We finally import LogisticRegression which is a training algorithm which will help us classify the dataset. We have made a variable 'clf' (classifier) and with this classifier we will train our model with x_train and y_train

```python
from sklearn.linear_model import LogisticRegression
clf = LogisticRegression()
```

```python
clf.fit(x_train, y_train)
```

```python
y_pred = clf.predict(x_test)
y_pred
```

```python
y_test
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test, y_pred)
```

# Importing plot_decision_regions from mlxtend.plotting
We use plot_decision_regions function to visualize what our model predicted and how it predicted

```python
from mlxtend.plotting import plot_decision_regions
```

```python
plot_decision_regions(x_train, y_train.values, clf=clf)
```
